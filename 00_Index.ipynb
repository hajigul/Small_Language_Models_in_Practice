# Small Language Models in Practice — Lecture Notes

*Companion notebooks to the book by **Haji Gul**.*

Code-first notes, one notebook per chapter, in reading order. Each notebook has a **Setup** cell, then alternating code and short notes that mirror the book.

Run chapters in order the first time — later ones reuse earlier ideas.

## Chapters

- **[Ch 1: Why Small Language Models](Chapter_01_Why_Small_Language_Models.ipynb)** — When small beats large, the local stack, and your first generation.
- **[Ch 2: Running Models Locally](Chapter_02_Running_Models_Locally.ipynb)** — transformers vs. Ollama, tokenization, sampling dials, streaming.
- **[Ch 3: Fine-Tuning with LoRA](Chapter_03_FineTuning_with_LoRA.ipynb)** — LoRA fine-tuning end to end: data, adapters, training, reuse.
- **[Ch 4: Retrieval-Augmented Generation](Chapter_04_RetrievalAugmented_Generation.ipynb)** — Chunk, embed, store in LanceDB, retrieve, and answer from your docs.
- **[Ch 5: Agents and Tool Use](Chapter_05_Agents_and_Tool_Use.ipynb)** — The tool-calling loop from scratch, then via a framework.
- **[Ch 6: Quantization for Small Hardware](Chapter_06_Quantization_for_Small_Hardware.ipynb)** — 4-bit loading with bitsandbytes and GGUF for llama.cpp/Ollama.
- **[Ch 7: Serving and Deployment](Chapter_07_Serving_and_Deployment.ipynb)** — FastAPI service with streaming, plus the Ollama no-code path.
- **[Ch 8: Capstone: A Local Document Assistant](Chapter_08_Capstone.ipynb)** — Quantized model + RAG + tools + serving in one local assistant.
